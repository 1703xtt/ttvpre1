## Ayuda Memoria

### df_ventas['precio'] = df_ventas['precio'].str.replace('$', '', regex=False)

### df_ventas['precio']:

Selecciona la columna precio, que inicialmente se leyó como texto (object / string) porque contenía el símbolo de moneda (por ejemplo: "$1500.50").

### .str:

Es el accesor de cadenas de Pandas. Le indica a Pandas: "voy a aplicar una función de texto fila por fila a cada valor de esta columna". Sin .str, Pandas no sabría cómo manipular el texto dentro de una Serie.

### .replace('$', '', regex=False):

'$': El texto que querés buscar.

'': El texto por el cual lo reemplazás (en este caso, nada / vacío, es decir, lo estás borrando).

### regex=False: 

En expresiones regulares (regex), el símbolo $ significa "final de la línea". Al  poner  regex=False, le aclarás a Pandas que trate al $ como un carácter literal, un signo pesos común y corriente, evitando errores o comportamientos inesperados.

### df_ventas['precio'] = ...:

Sobrescribe la columna original con los nuevos valores limpios (por ejemplo, transforma "$1500.50" en "1500.50").

### la columna 

sigue siendo texto (solo que ahora no tiene el signo $). Para poder hacer cuentas matemáticas (como calcular total_venta), necesitás convertirla a número flotante con .astype(float) o pd.to_numeric()

## Ayuda Memoria


### df_ventas['precio'] = pd.to_numeric(df_ventas['precio'], errors='coerce')

### Esta línea convierte la columna precio a un tipo numérico (float o int), pero con una red de seguridad fundamental para evitar que el programa se rompa si encuentra valores extraños.


### pd.to_numeric(...)
Es una función nativa de Pandas diseñada específicamente para transformar datos a formato numérico. A diferencia de .astype(float), esta función está pensada para manejar columnas "sucias" o con datos inconsistentes.

### errors='coerce' (La clave de la línea)
El parámetro errors define qué debe hacer Pandas si se topa con un dato que no se puede convertir a número (por ejemplo: un espacio en blanco " ", un texto como "N/A", "gratis", o un símbolo que no se haya limpiado).

Pandas tiene tres comportamientos posibles:

### errors='raise' (por defecto): 
Si hay un solo valor inválido, corta el programa y lanza un error.

### errors='ignore': 
Si falla, deja la columna tal como estaba (en formato texto) sin avisar.

### errors='coerce' (forzar): 
Si un valor no se puede transformar a número, lo convierte automáticamente en NaN (nulo o valor faltante) en lugar de romper el código.

### ¿Por qué lo usamos?
Robustez: Si en alguna fila había un dato corrupto o mal escrito, no detiene la ejecución del notebook.

Facilita la auditoría: Todos los valores problemáticos quedan unificados como NaN.

Prepara el terreno para el paso siguiente: Luego podés usar df_ventas['precio'].isna().sum() para contar cuántos datos fallaron y tratarlos limpiamente (por ejemplo, eliminándolos con .dropna(), como se hizo con las 2 filas descartadas).

# Etapa 2: Preprocesamiento y Limpieza de datos 
## Actividad 1: Limpieza de datos

```python
# 1. Eliminación de filas duplicadas en df_ventas
df_ventas = df_ventas.drop_duplicates()

print("Filas y columnas tras eliminar duplicados:", df_ventas.shape)
print("Duplicados restantes:", df_ventas.duplicated().sum())
```

```python
# Quitar el signo $ y convertir a número decimal
df_ventas['precio'] = df_ventas['precio'].astype(str).str.replace('$', '', regex=False)
df_ventas['precio'] = pd.to_numeric(df_ventas['precio'], errors='coerce')

# Verificamos
df_ventas[['producto', 'precio']].head()
```

```python
# 1. Tratar nulos remanentes (las 2 filas sin precio/cantidad identificadas)
df_ventas = df_ventas.dropna(subset=['precio', 'cantidad'])

# 2. Asegurar tipo de datos numérico en cantidad y calcular total_venta
df_ventas['cantidad'] = df_ventas['cantidad'].astype(int)

df_ventas['total_venta'] = (df_ventas['precio'] * df_ventas['cantidad']).round(2)

# Verificación inmediata
print("Dimensiones finales de ventas limpias:", df_ventas.shape)
df_ventas[['producto', 'precio', 'cantidad', 'total_venta']].head(3)
```

### Documentación del proceso y resultados (Actividad 1)
- **Eliminación de duplicados**: Se auditaron y eliminaron 35 filas duplicadas en el conjunto de ventas mediante `drop_duplicates()`.
- **Limpieza de caracteres no deseados**: Se eliminó el símbolo `$` de la columna `precio` con `.str.replace('$', '', regex=False)` y se transformó a tipo numérico con `pd.to_numeric()`.
- **Tratamiento de nulos**: Se descartaron las filas con valores faltantes mediante `.dropna()`.
- **Variable calculada**: Se generó la métrica `total_venta = precio * cantidad`.
- **Resultado final**: Un conjunto de ventas limpio con 2998 registros listos para el análisis.

## Actividad 2: Transformación de datos
### 2.1 Identificación y filtrado de ventas con alto rendimiento

```python
# 1. Definir el umbral de alto rendimiento (ventas superiores al promedio general)
# promedio_venta = df_ventas['total_venta'].mean()
```

### Por sugerencia del Instructor se paso del promedio al quartil 75% [umbral de alto rendimiento]

```python

umbral_alto_rendimiento = df_ventas['total_venta'].quantile(0.75)

# 2. Filtrar transacciones de alto rendimiento usando filtro condicional
df_alto_rendimiento = df_ventas[df_ventas['total_venta'] >= umbral_alto_rendimiento].reset_index(drop=True)

# 3. Mostrar resumen del filtro y las primeras filas
print(f"Umbral de alto rendimiento: ${umbral_alto_rendimiento:.2f}")
print(f"Total de registros de alto rendimiento: {len(df_alto_rendimiento)}")
display(df_alto_rendimiento[['id_venta', 'producto', 'categoria', 'total_venta']].head(3))
```

## Actividad 3: Agregación
### 3.1 Resumen de ingresos generados por categoría de producto

```python
# Agrupar las ventas por categoría y calcular la suma total de ingresos 
ventas_por_categoria = df_ventas.groupby('categoria')['total_venta'].sum().reset_index()

# Ordenar de mayor a menor y formatear presentación
ventas_por_categoria = ventas_por_categoria.sort_values(by='total_venta', ascending=False)
ventas_por_categoria['total_venta'] = ventas_por_categoria['total_venta'].round(2)

print("Ingresos acumulados por categoría:")
display(ventas_por_categoria)
```

```python
# 1. Resumir ventas e inversión de marketing por producto 
resumen_ventas = df_ventas.groupby('producto')['total_venta'].sum().reset_index()
# display(f"Resumen Ventas: {resumen_ventas}")

print("Ventas")
display(df_ventas.head())

print("Resumen Ventas")
display(resumen_ventas.head())

resumen_mkt = df_marketing.groupby('producto')['costo'].sum().reset_index()
# display(f"Resumen Marketing: {resumen_mkt}")
print("Marketing")
display(df_marketing.head())
print("Resumen Marketing")
display(resumen_mkt.head())

```

```python
# 2. Integrar ambos DataFrames usando pd.merge() con inner join (Clase 08)
df_integrado = pd.merge(resumen_ventas, resumen_mkt, on='producto', how='inner')

# 3. Renombrar columnas para claridad y mostrar los primeros resultados
df_integrado.columns = ['Producto', 'Ingresos_Totales', 'Inversion_Marketing']
df_integrado = df_integrado.sort_values(by='Ingresos_Totales', ascending=False).reset_index(drop=True)

print("Resumen Ventas")
display(resumen_ventas.head())

print("Resumen Marketing")
display(resumen_mkt.head())

print("Integración Ventas/Marketing")
display(df_integrado.head(5))
```

### 4.2 Documentación, justificación metodológica y análisis de tendencias [Ver final del README]
- **Clave de integración (`on='producto'`)**: Se utilizó el nombre del producto como llave relacional por ser el identificador  presente en ambas fuentes de datos (operaciones comerciales y esfuerzos de difusión).
- **Tipo de combinación (`how='inner'`)**: Se seleccionó un *inner join* para garantizar una correspondencia exacta entre los productos que registraron inversión publicitaria y aquellos que efectivamente concretaron transacciones de venta, evitando la inserción de registros incompletos o valores nulos (`NaN`).
- **Análisis de tendencias observadas**:
  - Se verifica que productos como **Lámpara de mesa** ($ 82276.38) y **Auriculares** ($ 74175.58) lideran ampliamente los ingresos generados.
  - En publicidad invertimos prácticamente lo mismo para cada producto (entre $14 y $18). Sin embargo, la facturación final varió muchísimo entre ellos. Esto demuestra que los productos que más dinero generaron lo hicieron por su propio valor y por la demanda natural de los clientes, y no porque hayamos gastado más en promocionarlos.
